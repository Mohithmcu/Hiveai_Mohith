# 01. Exploratory Data Analysis — Customer Support on Twitter (SpotifyCares)

This notebook inspects the raw dataset `twcs.csv`, filters for `@SpotifyCares` interactions, and analyzes message characteristics (turn lengths, distributions, response times).

```python
import pandas as pd
import numpy as np
from pathlib import Path

RAW_CSV = Path("../data/raw/twcs.csv")
print(f"Checking {RAW_CSV} (exists: {RAW_CSV.exists()})")
```

```python
# Load sample or full dataset
df_sample = pd.read_csv(RAW_CSV, nrows=500_000, low_memory=False)
print(f"Loaded sample of {len(df_sample):,} rows")
df_sample.head()
```

```python
# Check top brands by customer support volume
brand_counts = df_sample[df_sample['inbound'] == False]['author_id'].value_counts()
print("Top Brands in first 500k rows:")
print(brand_counts.head(10))
```

```python
# Filter for SpotifyCares
spotify_replies = df_sample[df_sample['author_id'].str.lower() == 'spotifycares']
print(f"SpotifyCares replies in sample: {len(spotify_replies):,}")
spotify_replies[['tweet_id', 'text', 'created_at']].head()
```
